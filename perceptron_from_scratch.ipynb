```python
import numpy as np

# activation funcation (sigmoid)
def activation(z):
    return 1 / (1 + np.exp(-z))

# Derivative activation funcation for backprop
def activation_derivative(z):
    return activation(z) * (1 - activation(z))
```

```python
# Perceptron and apply activation funcation (feedforward)
def perceptron(inputs, weights, bias):
    weighted_sum = np.dot(inputs, weights) + bias
    return activation(weighted_sum)
```

```python
# compute loss (mse ==> Mean Square error)
def compute_loss(y_pred, y_true):
    return np.mean((y_pred - y_true) ** 2)
```

```python
def update_weight(inputs, weights, bias, y_pred, y_true, learning_rate = 0.01):
    # dLoss/dz for MSE with sigmoid: 2*(y_pred - y_true)/n * sigmoid'(z), where sigmoid'(z) = y_pred*(1-y_pred)
    delta = 2 * (y_pred - y_true) / len(y_true) * y_pred * (1 - y_pred)
    weights_gradient = np.dot(inputs.T, delta) # wiehts gradient
    bias_gradient = np.sum(delta)  # bais gredient

    # update weights and bais
    weights -= learning_rate * weights_gradient
    bias -= learning_rate * bias_gradient

    return weights, bias
```

```python
np.random.seed(42)
inputs = np.random.randint(0, 10, (3,2))  # 3, 2
weights = np.random.rand(2)
bias = np.random.rand(1)
y_true = np.array([1,1,0])
```

```python
for epoch in range(100):
    y_pred = perceptron(inputs, weights, bias)  # forwad
    loss = compute_loss(y_pred, y_true) # compute the loss
    weights, bias = update_weight(inputs, weights, bias, y_pred, y_true, learning_rate = 0.01)
    print(f"loss for epoch {epoch} and the loss is {loss}")
```
