```python
import seaborn as sns
df = sns.load_dataset('titanic')
df
```

```python
df.head(5)
```

```python
df.describe()
```

```python
# - select data with male in sex and store in male_data
cond = df['sex'] == 'male'
male_data = df[cond]
```

```python
#- groubby for each embark_town sum of survived
df.groupby('embark_town')['survived'].sum()
```

```python
# - mean fare for each sex
df.groupby('sex')['fare'].mean()
```

```python
# - make new column called family_size which is sibsp + parch
df['family_size']  = df['sibsp'] + df['parch']
df
```

```python
# - sort df by family_size ascending = True
df.sort_values('family_size', ascending = True, inplace=True)
#df['family_size'].sort_values(ascending=True)
```

```python
# - value count each sex
df['sex'].value_counts()
```

```python
# - value count each embark_town
df['embark_town'].value_counts()
```

```python
# - drop deck column
df.drop('deck',axis=1, inplace=True)
```

```python
# - check nan values
df.isnull().sum()
```

```python
# - fill missing values in column with right method (mean, median, mode)
age_mean = df['age'].mean()
df['age'].fillna(age_mean, inplace=True)
```

```python
embark_town_mode = df['embark_town'].mode()[0]
df['embark_town'].fillna(embark_town_mode, inplace=True)
```
